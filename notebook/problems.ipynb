### Exercise 1

We will use NumPy to obtain information to describe statistically.

- Generate an array of 100 elements following a normal distribution.
- Generate an array of 100 elements following a chi-square distribution with 3 degrees of freedom.
- Calculate the main metrics and statistical measures that best describe the two vectors.

````python
# TODO
import numpy as np
import statistics as stats
from scipy.stats import skew
from scipy.stats import kurtosis


normal_distribution = np.random.normal(size = 100)
chi_distribution = np.random.chisquare(5, 100)

print(f'Normal distribution: {normal_distribution} \n')
print(f'Chi-square distribution: {chi_distribution} \n')
````

````python

#
#   CENTRAL TENDENCY
#
# MEAN
print("MEAN")
print(f'Normal mean: {stats.mean(normal_distribution)}')
print(f'Chi mean: {stats.mean(chi_distribution)} \n')

# MEDIAN
print(f'Normal median: {stats.median(normal_distribution)}')
print(f'Chi median: {stats.median(chi_distribution)} \n')

# MODE
print(f'Normal mode: {stats.mode(normal_distribution)}')
print(f'Chi mode: {stats.mode(chi_distribution)} \n')

#
#   DISPERSION
#
# RANGE
print("RANGE")
normal_range = max(normal_distribution) - min(normal_distribution)
print(f'Normal range: {normal_range}')
chi_range = max(chi_distribution) - min(chi_distribution)
print(f'Chi range: {chi_range} \n')

# VARIANCE AND STANDARD DEVIATION
print("VARIANCE AND STANDARD DEVIATION")
print(f'Normal variance: {stats.variance(normal_distribution)}')
print(f'Normal standard deviation: {stats.stdev(normal_distribution)}')
print(f'Chi variance: {stats.variance(chi_distribution)}')
print(f'Chi standard deviation: {stats.stdev(chi_distribution)} \n')

#
#   SHAPE MEASURES
#
# SKEWNESS
print("SKEWNESS")
normal_skewness = skew(normal_distribution)
print(f'Normal skewness: {normal_skewness}')
chi_skewness = skew(chi_distribution)
print(f'Chi skewness: {chi_skewness} \n')

# KURTOSIS
print("KURTOSIS")
normal_kurtosis = skew(normal_distribution)
print(f'Normal kurtosis: {normal_kurtosis}')
chi_kurtosis = skew(chi_distribution)
print(f'Chi kurtosis: {chi_kurtosis} \n')
````

### Exercise 2

Write a Python program to calculate the standard deviation of the following data:

```py
data = [4, 2, 5, 8, 6]
```

````python
# TODO
import statistics as stats
import math
data = [4, 2, 5, 8, 6]

def calc_stdev(data):
    
    x = len(data)
    stdev = 0
    mean = 0

    for num in data:
        mean += num
    
    mean = mean / x

    for num in data:
        stdev += (num - mean) ** 2
    stdev = math.sqrt(stdev / (num - 1))
    return stdev

print(f'Standard deviation: {calc_stdev(data)}')
'''std = stats.stdev(data)

print(f'Standard deviation: {round(std, 2)}')'''
````
